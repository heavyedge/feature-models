# Quantile crossing of direct GPQR models

```python
import sys
import os
import warnings

import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

## Load data

```python
crossing_H = pd.read_csv("../../_temp/v0/crossing.DirectMTGPQR_H.csv")
crossing_phi = pd.read_csv("../../_temp/v0/crossing.DirectMTGPQR_phi.csv")
```

## Plot

## H

```python
plt.plot(crossing_H["epoch"], crossing_H["crossing_rate"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Crossing Rate")
plt.show()
```

```python
plt.plot(crossing_H["epoch"], crossing_H["mean_crossing"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Mean crossing")
plt.show()
```

```python
plt.plot(crossing_H["epoch"], crossing_H["max_crossing"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Max crossing")
plt.show()
```

## phi

```python
plt.plot(crossing_phi["epoch"], crossing_phi["crossing_rate"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Crossing Rate")
plt.show()
```

```python
plt.plot(crossing_phi["epoch"], crossing_phi["mean_crossing"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Mean crossing")
plt.show()
```

```python
plt.plot(crossing_phi["epoch"], crossing_phi["max_crossing"])
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Max crossing")
plt.show()
```
