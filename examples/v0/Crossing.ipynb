# Quantile crossing of direct GPQR models

```python
import sys
import os
import warnings

import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

## Load data

```python
crossing_H = pd.read_csv("../../_temp/v0/crossing.DirectMTGPQR_H.csv")
crossing_phi = pd.read_csv("../../_temp/v0/crossing.DirectMTGPQR_phi.csv")
```

## Plot

## H

```python
target1 = crossing_H[crossing_H["test"] == 1]
target2 = crossing_H[crossing_H["test"] == 2]
```

```python
plt.plot(target1["epoch"], target1["crossing_rate"], label="Target 1")
plt.plot(target2["epoch"], target2["crossing_rate"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Crossing Rate")
plt.legend()
plt.show()
```

```python
plt.plot(target1["epoch"], target1["mean_crossing"], label="Target 1")
plt.plot(target2["epoch"], target2["mean_crossing"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Mean crossing")
plt.legend()
plt.show()
```

```python
plt.plot(target1["epoch"], target1["max_crossing"], label="Target 1")
plt.plot(target2["epoch"], target2["max_crossing"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.title("Max crossing")
plt.legend()
plt.show()
```
