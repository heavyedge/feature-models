# Quantile crossing of direct GPQR models

```python
import sys
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

# H

```python
direct_gpqr_independent = pd.read_csv(
    "../../benchmarks/v0/quantile_crossing.H.direct_gpqr_independent.csv",
    index_col=["index", "q1", "q2", "target"],
)
direct_gpqr_lmc = pd.read_csv(
    "../../benchmarks/v0/quantile_crossing.H.direct_gpqr_lmc.csv",
    index_col=["index", "q1", "q2", "target"],
)
```

## Crossing rate

```python
data = [
    (
        "Independent direct GPQR",
        direct_gpqr_independent["crossing_rate"].groupby(level=["q1", "q2"]).mean(),
    ),
    (
        "LMC direct gpqr",
        direct_gpqr_lmc["crossing_rate"].groupby(level=["q1", "q2"]).mean(),
    ),
]

plt.figure(figsize=(8, 4))

bar_width = 0.8 / len(data)
x = np.arange(len(data[0][1]))

for i, (label, df) in enumerate(data):
    offset = (i - (len(data) - 1) / 2) * bar_width
    plt.bar(x + offset, df.values, width=bar_width, label=label)

plt.legend()
plt.xlabel("Quantile pair")
plt.ylabel("Average crossing rate")
plt.title("Crossing rate")
plt.xticks(x, [f"{q1}-{q2}" for q1, q2 in data[0][1].index], rotation=45)
plt.show()
```

## Mean value of crossing magnitude

```python
data = [
    (
        "Independent direct GPQR",
        direct_gpqr_independent["mean_crossing"].groupby(level=["q1", "q2"]).mean(),
    ),
    (
        "LMC direct gpqr",
        direct_gpqr_lmc["mean_crossing"].groupby(level=["q1", "q2"]).mean(),
    ),
]

plt.figure(figsize=(8, 4))

bar_width = 0.8 / len(data)
x = np.arange(len(data[0][1]))

for i, (label, df) in enumerate(data):
    offset = (i - (len(data) - 1) / 2) * bar_width
    plt.bar(x + offset, df.values, width=bar_width, label=label)

plt.legend()
plt.xlabel("Quantile pair")
plt.ylabel("Average of mean crossing value")
plt.title("Mean crossing")
plt.xticks(x, [f"{q1}-{q2}" for q1, q2 in data[0][1].index], rotation=45)
plt.show()
```
