# Quantile crossing of direct GPQR models

```python
import sys
import os
import warnings

import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

# H

```python
direct_gpqr = pd.read_csv(
    "../../benchmarks/v0/quantile_crossing.H.direct_gpqr.csv",
    index_col=["index", "q1", "q2", "target"],
)
```

## Crossing rate

```python
crossing_rate_mean = direct_gpqr["crossing_rate"].groupby(level=["q1", "q2"]).mean()
labels = [f"{q1}-{q2}" for q1, q2 in crossing_rate_mean.index]

plt.figure(figsize=(8, 4))
plt.bar(labels, crossing_rate_mean.values, label="Direct GPQR")
plt.legend()
plt.xlabel("Quantile pair")
plt.ylabel("Average crossing rate")
plt.title("Crossing rate")
plt.xticks(rotation=45)
plt.show()
```

## Mean value of crossing magnitude

```python
mean_crossing_mean = direct_gpqr["mean_crossing"].groupby(level=["q1", "q2"]).mean()
labels = [f"{q1}-{q2}" for q1, q2 in mean_crossing_mean.index]

plt.figure(figsize=(8, 4))
plt.bar(labels, mean_crossing_mean.values, label="Direct GPQR")
plt.legend()
plt.xlabel("Quantile pair")
plt.ylabel("Average of mean crossing value")
plt.title("Mean crossing")
plt.xticks(rotation=45)
plt.show()
```

## Maximum value of crossing magnitude

```python
max_crossing_mean = direct_gpqr["max_crossing"].groupby(level=["q1", "q2"]).mean()
labels = [f"{q1}-{q2}" for q1, q2 in max_crossing_mean.index]

plt.figure(figsize=(8, 4))
plt.bar(labels, max_crossing_mean.values, label="Direct GPQR")
plt.legend()
plt.xlabel("Quantile pair")
plt.ylabel("Average of max crossing value")
plt.title("Max crossing")
plt.xticks(rotation=45)
plt.show()
```
