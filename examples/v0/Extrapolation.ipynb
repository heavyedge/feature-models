# Extrapolation of GPQR models

This notebook visualizes extrapolation performance of models with and without physical priors.

Models with constant prior show inferior extrapolation performance.

```python
import sys
import os
import math
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

## Load data

```python
H_cg = pd.read_csv("../../_temp/v0/extrapolation.CenterGapMTGPQR_H.csv")
H_cgcm = pd.read_csv("../../_temp/v0/extrapolation.CenterGapMTGPQR_H_ConstantMean.csv")
```

## Plot

```python
H_cg_loss = np.mean(
    [df["test_pinball_loss"] for (_, df) in H_cg.groupby("fold")], axis=0
)
H_cgcm_loss = np.mean(
    [df["test_pinball_loss"] for (_, df) in H_cgcm.groupby("fold")], axis=0
)
```

```python
plt.plot(np.arange(1, len(H_cg_loss) + 1), H_cg_loss, label="Informative mean")
plt.plot(np.arange(1, len(H_cgcm_loss) + 1), H_cgcm_loss, label="Constant mean")
plt.xlabel("Epoch")
plt.ylabel("Pinball loss")
plt.title("Extrapolation loss by epoch")
plt.legend()
plt.show()
```
