# Model evaluation

```python
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")
```

## Mean models

```python
gpr_indep_rmse = (
    pd.read_csv("../../benchmarks/v2/rmse.gpr_independent.csv", index_col=0)
    .groupby("target")
    .mean()
)
gpr_lmc_rmse = (
    pd.read_csv("../../benchmarks/v2/rmse.gpr_lmc.csv", index_col=0)
    .groupby("target")
    .mean()
)
gpr_indep_nlpd = (
    pd.read_csv("../../benchmarks/v2/nlpd.gpr_independent.csv", index_col=0)
    .groupby("target")
    .mean()
)
gpr_lmc_nlpd = (
    pd.read_csv("../../benchmarks/v2/nlpd.gpr_lmc.csv", index_col=0)
    .groupby("target")
    .mean()
)
```

```python
target = "H"
metrics = ["RMSE", "-NLPD"]
indep_values = [
    gpr_indep_rmse.loc[target, "rmse"],
    -gpr_indep_nlpd.loc[target, "nlpd"],
]
lmc_values = [
    gpr_lmc_rmse.loc[target, "rmse"],
    -gpr_lmc_nlpd.loc[target, "nlpd"],
]

x = np.arange(len(metrics))
bar_width = 0.35
fig, ax_rmse = plt.subplots(figsize=(6, 4))
ax_nlpd = ax_rmse.twinx()

ax_rmse.bar(
    x[0] - bar_width / 2, indep_values[0], bar_width,
    label="Independent", color="tab:blue",
)
ax_rmse.bar(
    x[0] + bar_width / 2, lmc_values[0], bar_width,
    label="LMC", color="tab:orange",
)
ax_nlpd.bar(
    x[1] - bar_width / 2, indep_values[1], bar_width, color="tab:blue",
)
ax_nlpd.bar(
    x[1] + bar_width / 2, lmc_values[1], bar_width, color="tab:orange",
)

ax_rmse.set_xticks(x, metrics)
ax_rmse.set_ylabel("RMSE")
ax_nlpd.set_ylabel("-NLPD")
ax_rmse.set_title(f"GPR metrics ({target})")
ax_rmse.legend()
fig.tight_layout()
plt.show()
```

## Quantile models

```python
gpr_indep_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpr_independent.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpr_lmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpr_lmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
```

```python
gpqr_indep_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_independent.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpqr_lmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_lmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpqr_cglmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_cglmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
```

```python
data = [
    (
        "GPR (Indep.)",
        gpr_indep_pinball[gpr_indep_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPR (LMC)",
        gpr_lmc_pinball[gpr_lmc_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (Indep.)",
        gpqr_indep_pinball[gpqr_indep_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (LMC)",
        gpqr_lmc_pinball[gpqr_lmc_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (Center-Gap LMC)",
        gpqr_cglmc_pinball[gpqr_cglmc_pinball.index.get_level_values("target") == "H"],
    ),
]

fig, ax = plt.subplots(figsize=(10, 4))

bar_width = 0.8 / len(data)
quantile_levels = data[0][1].index.get_level_values("quantile_level")
x = np.arange(len(quantile_levels))

for i, (label, df) in enumerate(data):
    offset = (i - (len(data) - 1) / 2) * bar_width
    ax.bar(x + offset, df["loss"], width=bar_width, label=label)

ax.set_xlabel("Quantile level")
ax.set_ylabel("Pinball loss")
ax.set_title("Mean pinball loss by quantile level (H)")
ax.set_xticks(x, quantile_levels.astype(str), rotation=45)
ax.legend()
fig.tight_layout()
plt.show()
```
