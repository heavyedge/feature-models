# Model evaluation

```python
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")
```

## Mean models

## Root mean squared error of latent function

```python
gpr_indep_rmse = (
    pd.read_csv("../../benchmarks/v2/rmse.gpr_independent.csv", index_col=0)
    .groupby("target")
    .mean()
)
gpr_lmc_rmse = (
    pd.read_csv("../../benchmarks/v2/rmse.gpr_lmc.csv", index_col=0)
    .groupby("target")
    .mean()
)
```

```python
data = [
    ("GPR (Independent)", gpr_indep_rmse),
    ("GPR (LMC)", gpr_lmc_rmse),
]

plot_data = pd.concat(
    [df.assign(model=label).reset_index() for label, df in data],
    ignore_index=True,
)

targets = plot_data["target"].unique()
fig, axes = plt.subplots(1, len(targets), figsize=(4 * len(targets), 4))

for ax, (target, df) in zip(axes, plot_data.groupby("target", sort=False)):
    ax.bar(df["model"], df["rmse"])
    ax.set_title(target)
    ax.set_xlabel("Model")
    ax.tick_params(axis="x", rotation=45)

axes[0].set_ylabel("RMSE")
fig.suptitle("Mean RMSE by target")
fig.tight_layout()
plt.show()
```

## Quantile models

```python
gpr_indep_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpr_independent.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpr_lmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpr_lmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
```

```python
gpqr_indep_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_independent.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpqr_lmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_lmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
gpqr_cglmc_pinball = (
    pd.read_csv("../../benchmarks/v2/pinball_loss.gpqr_cglmc.csv", index_col=0)
    .groupby(["target", "quantile_level"])
    .mean()
)
```

```python
data = [
    (
        "GPR (Indep.)",
        gpr_indep_pinball[gpr_indep_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPR (LMC)",
        gpr_lmc_pinball[gpr_lmc_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (Indep.)",
        gpqr_indep_pinball[gpqr_indep_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (LMC)",
        gpqr_lmc_pinball[gpqr_lmc_pinball.index.get_level_values("target") == "H"],
    ),
    (
        "GPQR (Center-Gap LMC)",
        gpqr_cglmc_pinball[gpqr_cglmc_pinball.index.get_level_values("target") == "H"],
    ),
]

fig, ax = plt.subplots(figsize=(10, 4))

bar_width = 0.8 / len(data)
quantile_levels = data[0][1].index.get_level_values("quantile_level")
x = np.arange(len(quantile_levels))

for i, (label, df) in enumerate(data):
    offset = (i - (len(data) - 1) / 2) * bar_width
    ax.bar(x + offset, df["loss"], width=bar_width, label=label)

ax.set_xlabel("Quantile level")
ax.set_ylabel("Pinball loss")
ax.set_title("Mean pinball loss by quantile level (H)")
ax.set_xticks(x, quantile_levels.astype(str), rotation=45)
ax.legend()
fig.tight_layout()
plt.show()
```
