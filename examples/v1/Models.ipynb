# Visualization of trained models

```python
import sys
import os
import warnings

import numpy as np
import torch
import pandas as pd

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")

torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SLURRIES = ["G50", "G45", "G40", "G40+IPA"]
```

```python
X = pd.read_csv("../../_temp/v1/X.csv", index_col=[0, 1, 2])
y = pd.read_csv("../../_temp/v1/y.csv", index_col=[0, 1, 2])
Xpred = pd.read_csv("../../_temp/v1/Xpred_1D.csv", index_col=[0, 1, 2])
```

```python
columns = ["slurry", "cosine_of_contact_angle"]
cos_thetas = X["cosine_of_contact_angle"].reset_index()[columns]
slurry_map = cos_thetas.drop_duplicates().set_index("cosine_of_contact_angle")["slurry"]

slurries = Xpred["cosine_of_contact_angle"].map(slurry_map)
unique_slurries = [s for s in SLURRIES if s in slurries.unique()]
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.ticker import ScalarFormatter

Rgt_pred = Xpred["gap_to_thickness_ratio"].unique()
Cas = Xpred["capillary_number"].unique()

Cas_sorted = np.sort(Cas)
cmap = plt.get_cmap("viridis", len(Cas_sorted))
norm = mcolors.BoundaryNorm(
    np.concatenate(
        [
            [Cas_sorted[0] * 0.9],
            (Cas_sorted[:-1] + Cas_sorted[1:]) / 2,
            [Cas_sorted[-1] * 1.1],
        ]
    ),
    ncolors=len(Cas_sorted),
)
```

## Prior mean

```python
pred_mean = pd.read_csv("../../_temp/v1/prior_mean.Xpred_1D.csv").pivot(
    index=["index", "batch"], columns="target", values="value"
)
```

```python
fig, axes = plt.subplots(1, len(slurry_map), sharex=True, sharey="row")

for slurry, ax in zip(unique_slurries, axes):
    ok = X.index.get_level_values("slurry") == slurry
    this_X = X[ok]
    this_H = y[ok]["H"].values

    ok_pred = slurries == slurry
    this_Xpred = Xpred[ok_pred]
    this_mean = pred_mean["H"][ok_pred.values]

    for ca in this_X["capillary_number"].unique():
        ok = this_X["capillary_number"] == ca
        ok_pred = this_Xpred["capillary_number"] == ca

        x_obs = this_X[ok]["gap_to_thickness_ratio"].to_numpy()
        y_obs = this_H[ok]
        for x0 in np.unique(x_obs):
            y_group = y_obs[x_obs == x0]
            if len(y_group) > 0:
                q1 = np.quantile(y_group, 0.025)
                q3 = np.quantile(y_group, 0.975)
                ax.vlines(
                    x0,
                    ymin=q1,
                    ymax=q3,
                    color=cmap(norm(ca)),
                    linewidth=1.5,
                    alpha=0.7,
                )

        ax.plot(
            this_Xpred[ok_pred]["gap_to_thickness_ratio"],
            this_mean[ok_pred.values],
            color=cmap(norm(ca)),
        )

    (cos_theta,) = this_X["cosine_of_contact_angle"].unique()
    ax.set_title(f"Cos θ={cos_theta:.2f}")

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(
    sm, ax=axes, orientation="horizontal", location="top", pad=0.2, aspect=30
)
cbar.set_label("Ca")
quartile_vals = np.quantile(Cas_sorted, [0, 0.25, 0.5, 0.75, 1.0])
nearest_cas = [Cas_sorted[np.argmin(np.abs(Cas_sorted - q))] for q in quartile_vals]
cbar.set_ticks([round(ca, 3) for ca in nearest_cas])
cbar.ax.xaxis.set_major_formatter(ScalarFormatter())

fig.supxlabel("Rgt")
fig.supylabel("H")

fig.suptitle("Prior mean")
plt.show()
```

## GPR

```python
pred_gpr = pd.read_csv("../../_temp/v1/gpr.Xpred_1D.csv").pivot(
    index=["index", "batch"],
    columns="target",
)
```

```python
fig, axes = plt.subplots(1, len(slurry_map), sharex=True, sharey="row")

for slurry, ax in zip(unique_slurries, axes):
    ok = X.index.get_level_values("slurry") == slurry
    this_X = X[ok]
    this_H = y[ok]["H"].values

    ok_pred = slurries == slurry
    this_Xpred = Xpred[ok_pred]
    this_mean = pred_gpr["latent_mean"]["H"][ok_pred.values]
    this_std = pred_gpr["latent_std"]["H"][ok_pred.values]

    for ca in this_X["capillary_number"].unique():
        ok = this_X["capillary_number"] == ca
        ok_pred = this_Xpred["capillary_number"] == ca

        x_obs = this_X[ok]["gap_to_thickness_ratio"].to_numpy()
        y_obs = this_H[ok]
        for x0 in np.unique(x_obs):
            y_group = y_obs[x_obs == x0]
            if len(y_group) > 0:
                q1 = np.quantile(y_group, 0.025)
                q3 = np.quantile(y_group, 0.975)
                ax.vlines(
                    x0,
                    ymin=q1,
                    ymax=q3,
                    color=cmap(norm(ca)),
                    linewidth=1.5,
                    alpha=0.7,
                )

        ax.fill_between(
            this_Xpred[ok_pred]["gap_to_thickness_ratio"],
            this_mean[ok_pred.values] - this_std[ok_pred.values],
            this_mean[ok_pred.values] + this_std[ok_pred.values],
            color=cmap(norm(ca)),
            alpha=0.2,
        )

    (cos_theta,) = this_X["cosine_of_contact_angle"].unique()
    ax.set_title(f"Cos θ={cos_theta:.2f}")

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(
    sm, ax=axes, orientation="horizontal", location="top", pad=0.2, aspect=30
)
cbar.set_label("Ca")
quartile_vals = np.quantile(Cas_sorted, [0, 0.25, 0.5, 0.75, 1.0])
nearest_cas = [Cas_sorted[np.argmin(np.abs(Cas_sorted - q))] for q in quartile_vals]
cbar.set_ticks([round(ca, 3) for ca in nearest_cas])
cbar.ax.xaxis.set_major_formatter(ScalarFormatter())

fig.supxlabel("Rgt")
fig.supylabel("H")
fig.suptitle("Posterior distribution of mean")
plt.show()
```

## GPQR

```python
pred_gpqr = pd.read_csv("../../_temp/v1/gpqr.Xpred_1D.csv").pivot(
    index=["index", "batch", "quantile", "sample"],
    columns="target",
)

gpqr_levels = pred_gpqr.index.get_level_values("quantile").unique().sort_values()
lower_gpqr = (
    pred_gpqr["value"]
    .xs(gpqr_levels[0], level="quantile")
    .groupby(level="index")
    .quantile(0.025)
)
upper_gpqr = (
    pred_gpqr["value"]
    .xs(gpqr_levels[-1], level="quantile")
    .groupby(level="index")
    .quantile(0.975)
)
```

```python
fig, axes = plt.subplots(1, len(slurry_map), sharex=True, sharey="row")

for slurry, ax in zip(unique_slurries, axes):
    ok = X.index.get_level_values("slurry") == slurry
    this_X = X[ok]
    this_y = y.loc[ok, "H"].to_numpy()

    ok_pred = slurries == slurry
    this_Xpred = Xpred[ok_pred].copy()
    this_Xpred["prediction_index"] = np.flatnonzero(ok_pred)

    for ca in this_X["capillary_number"].unique():
        ok = this_X["capillary_number"] == ca
        ok_pred = this_Xpred["capillary_number"] == ca

        x_obs = this_X.loc[ok, "gap_to_thickness_ratio"].to_numpy()
        y_obs = this_y[ok]
        for x0 in np.unique(x_obs):
            y_group = y_obs[x_obs == x0]
            if len(y_group) > 0:
                q1, q3 = np.quantile(y_group, [0.025, 0.975])
                ax.vlines(
                    x0,
                    ymin=q1,
                    ymax=q3,
                    color=cmap(norm(ca)),
                    linewidth=1.5,
                    alpha=0.7,
                )

        prediction_index = this_Xpred.loc[ok_pred, "prediction_index"]
        ax.fill_between(
            this_Xpred.loc[ok_pred, "gap_to_thickness_ratio"],
            lower_gpqr.loc[prediction_index, "H"],
            upper_gpqr.loc[prediction_index, "H"],
            facecolor=cmap(norm(ca)),
            edgecolor="none",
            alpha=0.3,
        )

    (cos_theta,) = this_X["cosine_of_contact_angle"].unique()
    ax.set_title(f"Cos θ={cos_theta:.2f}")

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(
    sm, ax=axes, orientation="horizontal", location="top", pad=0.2, aspect=30
)
cbar.set_label("Ca")
quartile_vals = np.quantile(Cas_sorted, [0, 0.25, 0.5, 0.75, 1.0])
nearest_cas = [Cas_sorted[np.argmin(np.abs(Cas_sorted - q))] for q in quartile_vals]
cbar.set_ticks([round(ca, 3) for ca in nearest_cas])
cbar.ax.xaxis.set_major_formatter(ScalarFormatter())

fig.supxlabel("Rgt")
fig.supylabel("H")
fig.suptitle("Quantiles")
plt.show()
```
