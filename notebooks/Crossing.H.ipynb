# Quantile crossing validation of direct GPQR models for H

```python
import sys
import os
import warnings

import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

warnings.filterwarnings("ignore")
```

## Load data

```python
crossing = pd.read_csv("../_temp/crossing.DirectMTGPQR_H.csv")
```

## Plot

```python
plt.plot(crossing["crossing_rate_0"], label="Target 1")
plt.plot(crossing["crossing_rate_1"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Crossing Rate")
plt.legend()
plt.show()
```

```python
plt.plot(crossing["mean_crossing_0"], label="Target 1")
plt.plot(crossing["mean_crossing_1"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Mean crossing")
plt.legend()
plt.show()
```

```python
plt.plot(crossing["max_crossing_0"], label="Target 1")
plt.plot(crossing["max_crossing_1"], label="Target 2")
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Max crossing")
plt.legend()
plt.show()
```
